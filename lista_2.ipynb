### Zadanie 7

```python
A=np.matrix("10.0 -2 -1 2 3 1 -4 7;\
           5 11 3 10 -3 3 3 -4;\
           7 12 1 5 3 -12 2 3;\
           8 7 -2 1 3 2 2 4;\
           2 -15 -1 1 4 -1 8 3;\
           4 2 9 1 12 -1 4 1;\
           -1 4 -7 -1 1 1 -1 -3;\
           -1 3 4 1 3 -4 7 6")
B = np.matrix("0.0 12 -5 3 -25 -26 9 -7").transpose()

def gauss_solver(matrix_1,end_matrix):
    for i in range(matrix_1.shape[0]):
        other_rows=list(range(i))
        other_rows.extend(list(range(i+1,matrix_1.shape[0])))
        div=1/matrix_1[i,i]
        matrix_1[i]=matrix_1[i]*(1/matrix_1[i,i])
        end_matrix[i]=end_matrix[i]*(div)
        for j in other_rows:
            div=matrix_1[j,i]
            matrix_1[j]=matrix_1[j]-(div*matrix_1[i])
            end_matrix[j]=end_matrix[j]-(div*end_matrix[i])
    return end_matrix  
#szybki test na działanie metody
#=======
#ex1=np.matrix("2.0 1; 1 2")
#x2=np.matrix("1.0 1").transpose()
#gauss_solver(ex1,ex2)
#powinno wyjśc "1/3 ; 1/3"
#======
out=gauss_solver(A,B)
for i in range(out.shape[0]):
    print(f"x{1} = {out[i,0]}")
```
